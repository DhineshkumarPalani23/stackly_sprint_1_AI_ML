# Data Types

A data type tells Python what type of value is stored in a variable.

Python has different built-in data types for storing different kinds of data.

## Common Python Data Types

- `int` – used to store whole numbers.

- `float` – used to store decimal numbers.

- `str` – used to store text. A string can be enclosed within single quotes (`' '`) or double quotes (`" "`).

- `bool` – used to store `True` or `False` values.

```python
name="dhineshkumar"
age=22
height=5.5
is_employee=True
print(name,type(name))
print(age,type(age))
print(height,type(height))
print(is_employee,type(is_employee))
```

## Code Explanation

In this program, I created four variables: `name`, `age`, `height`, and `is_employee`.

- The `name` variable stores a string value.
- The `age` variable stores an integer value.
- The `height` variable stores a decimal value.
- The `is_employee` variable stores a Boolean value.

I used two functions: `print()` and `type()`.

- The `print()` function displays the values in the console.
- The `type()` function tells us the data type of a value.

# Type Casting

Type casting means converting a value from one data type to another data type

## Type casting built-in function

`int()`- converts a value into an integer.

`float()`- converts a value into a floating-point number

`bool()`- converts a value into a boolean values

`str()` - converts a value into a string value.`

# Operators

Operators is a symbols or keywords used to perform operations on value and variabes

## Types of Operators

- Arithmetic Operators
- Assignment Operators
- Comparison Operators
- Logical Operators

### Arithmetic Operators

Arithmetic operators are used to perform mathematical operations.

Examples: `+`, `-`, `*`, `/`, `%`, `//`, `**`

### Assignment Operators

Assignment operators are used to assign or update values in variables.

Examples: `=`, `+=`, `-=`, `*=`, `/=`

### Comparison Operators

Comparison operators are used to compare two values.

Examples: `==`, `!=`, `>`, `<`, `>=`, `<=`

### Logical Operators

Logical operators are used to combine or evaluate conditions.

Examples: `and`, `or`, `not`



```python
# Student marks
python_mark = 60
sql_mark = 80
math_mark = 90

# Calculate total marks
total = python_mark + sql_mark + math_mark

# Calculate average marks
average = total / 3

# Passing average
passing_mark = 50

# Check pass or fail
is_pass = average >= passing_mark

# Display the results
print("Python Mark:", python_mark)
print("SQL Mark:", sql_mark)
print("Math Mark:", math_mark)

print("Total Marks:", total)
print("Average Marks:", average)
print("Pass:", is_pass)

if is_pass:
    print("Result: Pass")
else:
    print("Result: Fail")
```
