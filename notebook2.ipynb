### Constructor __init__()

Constructor (__init__()) is a special method that automatically executes when an object is created.

It is mainly used to give the object its initial data.

Real-world example
    
Imagine a college creates a student record.

```python
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

student=Student("Dhinesh",21)

print(student.name)
print(student.age)
```

### Instance Variables

An instance variable is a variable that belongs to a specific object.

Real-world example
    
Imagine a Student class.

Each student can have a different:

-name

-age

-course

```python
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

student1=Student("Dhinesh",21)
student2=Student("Arun",22)

print(student1.name)
print(student1.age)

print(student2.name)
print(student2.age)
```

### class variable 

A class variable is a variable that belongs to the class itself. It is created inside the class and outside the methods. All objects of that class can access and share the class variable.

Real-world example:

All students belong to the same school.

```python
class Student:
    school="ABC School"       # Class variable

    def __init__(self,name,age):
        self.name=name        # Instance variable
        self.age=age          # Instance variable

student1=Student("Dhinesh",21)
student2=Student("Arun",22)

print(student1.name)
print(student1.age)
print(student1.school)

print(student2.name)
print(student2.age)
print(student2.school)
```

### Instance Method

An instance method is a method that belongs to an object of a class.

It is used to perform an action using the data of a particular object.

An instance method normally has self as its first parameter.

Real-world example:

A student can perform an action called study().

```python
class Student:
    def __init__(self,name):
        self.name=name

    def study(self):
        print(self.name,"is studying")

student1=Student("Dhinesh")

student1.study()
```

### Static Method

A static method is a method that belongs to a class but does not depend on a particular object.

We create a static method using the @staticmethod decorator.

Real-world example:

Suppose we want to check whether a mark is a pass or fail.

```python
class Student:
    @staticmethod
    def check_pass(mark):
        if mark>=50:
            print("Pass")
        else:
            print("Fail")

Student.check_pass(70)
Student.check_pass(40)
```

### Class Method?

A class method is a method that belongs to the class and works with class-level data.

We create a class method using the @classmethod decorator.

A class method takes cls as its first parameter.

Real-world example:

Suppose all students belong to the same school.

```python
class Student:
    school="ABC School"

    @classmethod
    def show_school(cls):
        print(cls.school)

Student.show_school()
```

### @Property

A property allows us to access a method like an ordinary variable.

We use the @property decorator to create a property.

Real-world example:

Student Full Name

```python
class Student:
    def __init__(self,first_name,last_name):
        self.first_name=first_name
        self.last_name=last_name

    @property
    def full_name(self):
        return self.first_name+" "+self.last_name

student=Student("Dhinesh","Kumar")

print(student.full_name)
```

### Encapsulation

Encapsulation means combining data and methods inside a class and controlling how the data can be accessed or modified.

Real-world example:

A bank account has sensitive balance information.

```python
class BankAccount:
    def __init__(self,balance):
        self.__balance=balance

    def get_balance(self):
        return self.__balance

account=BankAccount(10000)

print(account.get_balance())
```

### Abstraction

Abstraction means hiding unnecessary implementation details and showing only the important information to the user.

-ABC → Abstract Base Class
    
-@abstractmethod → defines a method that child classes must implement

Real-world example:

Import the abstract method using

    from abc import ABC,abstractmethod

Real-world example:
    
Every employee should have a work() method.

```python
from abc import ABC,abstractmethod

class Employee(ABC):

    @abstractmethod
    def work(self):
        pass


class Developer(Employee):

    def work(self):
        print("Developer is writing code")


employee=Developer()
employee.work()
```

### Inheritance

Inheritance means allowing a child class to access and reuse the properties and methods of a parent class.

Parent class → Existing class

Child class → New class that inherits from the parent

Inheritance → Relationship between parent and child classes

Real-world example:

Employee → Developer

```python
# Create the parent class
class Employee:

    # Create a method in the parent class
    def work(self):
        print("Employee is working")


# Create the child class and inherit from Employee
class Developer(Employee):

    # Create a method in the child class
    def write_code(self):
        print("Developer is writing code")


# Create an object of the child class
developer=Developer()

# Call the method inherited from the parent class
developer.work()

# Call the method of the child class
developer.write_code()
```

### Multiple Inheritance

Multiple inheritance is a type of inheritance where one child class inherits from two or more parent classes.

Real-world example:

Employee + Developer → AIEngineer

```python
class Employee:
    def work(self):
        print("Employee is working")


class Developer:
    def write_code(self):
        print("Developer is writing code")


# Multiple inheritance
class AIEngineer(Employee,Developer):
    def build_model(self):
        print("AI Engineer is building a model")


engineer=AIEngineer()

engineer=AIEngineer()
engineer.build_model()
engineer.write_code()
engineer.work()
```

### Multilevel Inheritance

Multilevel inheritance is a type of inheritance where a class inherits from another child class, creating multiple levels of inheritance.

Real-world example:

Example 1 – Employee → Developer → AIEngineer

```python
class Employee:
    def work(self):
        print("Employee is working")


class Developer(Employee):
    def write_code(self):
        print("Developer is writing code")


class AIEngineer(Developer):
    def build_model(self):
        print("AI Engineer is building a model")


engineer=AIEngineer()

engineer.work()
engineer.write_code()
engineer.build_model()
```

### Hierarchical Inheritance

Hierarchical inheritance is a type of inheritance where multiple child classes inherit from the same parent class.

Real-world example:

The Employee class is inherited by both Developer and Tester classes.

```python
class Employee:
    def work(self):
        print("Employee is working")


# Developer inherits from Employee
class Developer(Employee):
    def write_code(self):
        print("Developer is writing code")


# Tester inherits from Employee
class Tester(Employee):
    def test_application(self):
        print("Tester is testing the application")


developer=Developer()
tester=Tester()

developer.work()
developer.write_code()

tester.work()
tester.test_application()
```

### Polymorphism

Polymorphism means the same method or function name can perform different actions depending on the object or situation.

Real-world example:

Animals

Dog → sound() → Bark

Cat → sound() → Meow

```python
class Dog:
    def sound(self):
        print("Dog is barking")


class Cat:
    def sound(self):
        print("Cat is meowing")


dog=Dog()
cat=Cat()

dog.sound()
cat.sound()
```

### Overriding

Method overriding means a child class provides its own version of a method that is already defined in the parent class.

Real-world example:

Animals

Animal → sound() → Animal sound

Dog → sound() → Bark

The Dog class changes the sound() method of the Animal class.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog is barking")


dog=Dog()
dog.sound()
```

### Method overloading

Method overloading means using the same method name with different numbers or types of arguments.

Real-world example:

Addition

add(10,20)
    
add(10,20,30)

```python
class Calculator:
    def add(self,a,b=0,c=0):
        print(a+b+c)


calculator=Calculator()

calculator.add(10,20)
calculator.add(10,20,30)
```

### Composition

Composition means one class contains an object of another class and uses its properties and methods.

Real-world example:

Computer and CPU

A computer contains a CPU.

```python
class CPU:
    def process(self):
        print("CPU is processing")


class Computer:
    def __init__(self):
        self.cpu=CPU()  # Create CPU object inside Computer

    def start(self):
        self.cpu.process()  # Call CPU method
        print("Computer is running")


computer=Computer()
computer.start()
```

### Aggregation

Aggregation is an OOP concept where one class uses an object of another class, but the two objects can exist independently.

Real-world example:

A company has employees, but an employee can leave the company and still exist independently.

```python
class Employee:
    def work(self):
        print("Employee is working")


class Department:
    def __init__(self,employee):
        self.employee=employee

    def show_employee(self):
        self.employee.work()


employee=Employee()
department=Department(employee)

department.show_employee()
```

### Association

Association means a relationship between two classes where one class uses or interacts with another class.

Real-world example:

A customer contacts the support team to resolve a technical issue.

Customer ───── contacts ─────> Support Team

All three are Association because both sides exist independently and simply interact with each other.

```python
class SupportTeam:
    def solve_issue(self):
        print("Support team is solving the issue")


class Customer:
    def contact_support(self,support):
        support.solve_issue()  # Call SupportTeam method


support=SupportTeam()  # Create SupportTeam object
customer=Customer()  # Create Customer object

customer.contact_support(support)  # Customer interacts with SupportTeam
```

#### Explanation

SupportTeam → creates the Support Team class.
    
Customer → creates the Customer class.
    
support=SupportTeam() → creates a Support Team object.
    
customer=Customer() → creates a Customer object.
    
customer.contact_support(support) → Customer contacts the Support Team.
    
support.solve_issue() → Support Team solves the issue.

### Magic (Dunder) Methods

Magic methods, also called Dunder methods, are special methods in Python that start and end with double underscores (__).

Real-world example:

When an Employee object is created, Python can automatically call __init__() to initialize the employee's data.

```python
class Employee:
    def __init__(self,name):
        self.name=name  # Store employee name


employee=Employee("Dhinesh")  # Create Employee object

print(employee.name)  # Display employee name
```

super()

super() is used in a child class to access the parent class's methods or constructor.

Real-world example:

A child class can use a method that is already defined in its parent class.

```python
class Employee:
    def __init__(self,name):
        self.name=name


class AIEngineer(Employee):
    def __init__(self,name,skill):
        super().__init__(name)
        self.skill=skill


engineer=AIEngineer("Dhinesh","Python")

print(engineer.name)
print(engineer.skill)
```

#### Explanation
Employee is the parent class.

AIEngineer is the child class.

super().__init__(name) calls the __init__() method of the parent class.

self.skill=skill adds the AI Engineer's own information.

The program reuses the parent's name initialization instead of writing it again.

### __str__()

__str__() is a magic (dunder) method that defines what should be displayed when we use print() with an object.

Real-world example:

When we print an employee object, __str__() can display the employee's name in a clear and readable format.

```python
class Employee:
    def __init__(self,name):
        self.name=name

    def __str__(self):
        return "Employee: "+self.name


employee=Employee("Dhinesh")

print(employee)
```

#### Explanation

__init__() stores the employee's name.

__str__() defines the information to display.
    
print(employee) automatically calls __str__().

### __repr__()

__repr__() is a magic (dunder) method used to define the official or detailed string representation of an object.

Real-world example:

When a developer wants to inspect an Employee object, __repr__() can show the employee's name and ID clearly.

```python
class Employee:
    def __init__(self,name,employee_id):
        self.name=name
        self.employee_id=employee_id

    def __repr__(self):
        return f"Employee(name='{self.name}', employee_id={self.employee_id})"


employee=Employee("Dhinesh",101)

print(repr(employee))
```

#### Explanation

__init__() stores the employee's name and ID.

__repr__() defines how the object should be represented.
    
repr(employee) automatically calls __repr__().
    
It gives detailed information that is useful for developers.

### __len__()

__len__() is a magic (dunder) method used to define what should happen when we use the len() function with an object.

Real-world example:

Shopping Cart

When we check the length of a shopping cart, __len__() can return the number of products in the cart.

```python
class Project:
    def __init__(self,tasks):
        self.tasks=tasks

    def __len__(self):
        return len(self.tasks)


project=Project(["Python","SQL","OOP"])

print(len(project))
```

#### Explanation

__init__() stores the list of tasks.
    
__len__() tells Python how to calculate the project's length.

len(self.tasks) counts the number of tasks.
    
len(project) automatically calls __len__().

### Dataclasses (Bonus)

A dataclass is a special type of Python class mainly used to store data.

Real-world example:
    
An Employee dataclass can store an employee's name, ID, and designation.

```python
from dataclasses import dataclass

@dataclass
class Employee:
    name:str
    employee_id:int
    designation:str


employee=Employee("Dhinesh",101,"AI/ML Engineer")


print(employee.name)
print(employee.employee_id)
print(employee.designation)
```

#### Explanation

Imports dataclass from Python's dataclasses module.

@dataclass makes the Employee class a dataclass.
    
name, employee_id, and designation are the data fields.
    
We don't need to write __init__() manually.

Employee("Dhinesh",101,"AI/ML Engineer") creates the Employee object and stores the data

### Singleton Pattern (Bonus)

The Singleton Pattern is a design pattern that ensures a class has only one object (instance) throughout the program

Real-world example:
    
An application can use one shared database connection instead of creating multiple connection objects..

```python
class Database:
    instance=None

    def __new__(cls):
        if cls.instance is None:
            cls.instance=super().__new__(cls)
        return cls.instance


db1=Database()
db2=Database()

print(db1 is db2)
```

#### Explanation

Database is the Singleton class.
    
instance=None stores the single object.
    
__new__() controls object creation.
    
If an object does not exist, it creates one.
    
If an object already exists, it returns the same object.
    
db1 and db2 therefore refer to the same object.
