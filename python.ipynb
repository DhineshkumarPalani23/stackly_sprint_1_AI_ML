### Arithmetic Operators

Arithmetic operators are used to perform mathematical operations.

Examples: `+`, `-`, `*`, `/`, `%`, `//`, `**`

### Assignment Operators

Assignment operators are used to assign or update values in variables.

Logical operators are used to combine or evaluate conditions.

```python
# Student marks
python_mark=60
sql_mark=80
math_mark=90

# Calculate total marks
total=python_mark+sql_mark+math_mark

# Calculate average marks
average=total/3

# Passing average
passing_mark=50

# Check pass or fail
is_pass=average>=passing_mark

# Display the results
print("Python Mark:",python_mark)
print("SQL Mark:",sql_mark)
print("Math Mark:",math_mark)

print("Total Marks:",total)
print("Average Marks:",average)
print("Pass:",is_pass)

if is_pass:
    print("Result: Pass")
else:
    print("Result: Fail")
```

The break statement immediately stops the loop and exits it.


```python
for i in range(1,6):
    if i==3:
        break
    print(i)
```

The continue statement skips the current iteration and moves to the next iteration.

```python
for i in range(1,6):
    if i==4:
        continue
    print(i)
```

### Dictionary Comprehension

Dictionary comprehension is a short and simple way to create a new dictionary using a loop.

Normally, we use a for loop to create a dictionary. With dictionary comprehension, we can do the same thing in a single line.

## Step 2 – Demonstrate the Concept

```python
students=["Arun","Bala","Dhinesh"]
marks=[80,75,90]

student_marks={student:mark for student,mark in zip(students,marks)}

print(student_marks)
```

Keyword arguments are arguments that are passed to a function using the parameter name.

### Variable-Length Arguments (*args, **kwargs)

Variable-length arguments are used when we don't know how many arguments will be passed to a function.

-*args 

    *args is used when we want to pass any number of positional arguments.

-**kwargs 

    **kwargs is used when we want to pass any number of keyword arguments.

```python
#*args

def add_numbers(*numbers):
    total=0

    for number in numbers:
        total=total+number

    print("Total:",total)

add_numbers(10,20)
add_numbers(10,20,30)

print()

def student_details(**details):
    print(details)

student_details(name="Dhinesh",age=21,course="Python")
```

## Step 1 – Understand the Concept

-A recursive function is a function that calls itself to solve a problem.

-The function keeps calling itself with a smaller or simpler value until it reaches a stopping condition, called the base condition

-Recursive call → the function calls itself.

-Base condition → tells the function when to stop.

Real-time example: Countdown

Suppose we want to count down from 5:

```python
def countdown(number):
    if number==0:
        return

    print(number)
    countdown(number-1)

countdown(5)
```

## Generators

A generator is a special type of function that produces values one at a time instead of producing all values at once.

Generators use the yield keyword.

Real-time example:

Imagine an application has 1 lakh customer records.

Instead of loading all 1 lakh records into memory:


```python
def numbers():
    yield 1
    yield 2
    yield 3

result=numbers()

print(next(result))
print(next(result))
```

### yield

yield is a Python keyword used in a generator function.

It is used to produce one value at a time and pause the function.

### Demonstrate the Concept

Think of a vending machine:

Request → 1 item → pause
Request → 1 item → pause
Request → 1 item → pause


```python
def numbers():
    yield 10
    yield 20
    yield 30

result=numbers()

print(next(result))
print(next(result))
print(next(result))
```

### Decorators

A decorator is a function that is used to add extra functionality to another function without changing the original function's code.

Real-time example:

Imagine you have a basic mobile phone.

Basic Phone
    ↓
Add Camera
    ↓
Add extra feature

```python
def decorator(function):
    def wrapper():
        print("Before")
        function()
        print("After")
    return wrapper

@decorator
def hello():
    print("Hello")

hello()
```

### Closures

A closure is a function that remembers and uses a value from its outer function even after the outer function has finished running.


```python
def create_greeting(name):
    def greeting():
        print("Hello",name)
    return greeting

hello=create_greeting("Dhinesh")

hello()
```

### global

global is used inside a function when we want to change a variable that belongs to the global scope

```python
count=0

def increase():
    global count
    count=count+1

increase()

print(count)
```

### nonlocal

nonlocal is used inside a nested function when we want to change a variable that belongs to the outer function.


```python
def counter():
    count=0

    def increase():
        nonlocal count
        count=count+1
        print(count)

    increase()
    increase()

counter()
```
